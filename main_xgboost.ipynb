```python
#Merging stats and player data to get player position
stats_players=stats_data.merge(player_data[["PlayerId","Position"]], on="PlayerId", how="inner")
```

```python
stats_players.info()
```

```python
stats_players.head()
```

```python
#choosing numerical stats as features(per game averages):
feature_num=["Disposals", "Kicks", "Marks", "Handballs",
    "Goals", "Behinds", "HitOuts", "Tackles",
    "Rebounds", "Inside50s", "Clearances",
    "ContestedPossessions", "UncontestedPossessions",
    "ContestedMarks", "MarksInside50", "OnePercenters",
    "Bounces", "GoalAssists", "%Played"]
```

```python
#aggregate per player per game stats
player_features=(stats_players.groupby(['PlayerId','Position'])[feature_num].mean().reset_index())
```

```python
player_features.head()
```

```python
player_features.info()
```

```python
player_features['Position'].value_counts()
```

```python

def simplify_position(pos):
    if "Defender" in pos:
        return "Defender"
    if "Midfield" in pos:
        return "Midfield"
    if "Forward" in pos:
        return "Forward"
    if "Ruck" in pos:
        return "Ruck"
    return pos

player_features["Position_simple"] = player_features["Position"].apply(simplify_position)

X = player_features[feature_num]
y = player_features["Position_simple"]
labelencode=LabelEncoder()
y_encoded=labelencode.fit_transform(y)
X_train, X_test, y_train, y_test = train_test_split(X,y_encoded,test_size=0.2,random_state=42,stratify=y_encoded)
```

```python
from sklearn.preprocessing import StandardScaler
#scaling features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

```

```python
from sklearn.pipeline import _name_estimators
import xgboost as xgb
model_xgb=xgb.XGBClassifier(n_estimators=200,
    learning_rate=0.05,
    max_depth=5,
    objective='multi:softmax',
    num_class=4,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='mlogloss')
```

```python
model_xgb.fit(X_train_scaled, y_train)
```

```python
xgb_y_pred=model_xgb.predict(X_test_scaled)
```

```python
print("XGBoost Accuracy:", accuracy_score(y_test, xgb_y_pred))
print(classification_report(y_test, xgb_y_pred, target_names=labelencode.classes_))
print(confusion_matrix(y_test, xgb_y_pred))
```
